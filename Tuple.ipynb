```python
#tuple
#important point : Parentheses() are commonly used to create tuple
data=(10,20,30,40,50)
print(type(data))
```

```python
#but prentheses are not actually mandatory
#the comma, is what makes a tuple, not the parenthese.
data2=10,20,30,40,50
print(type(data))
```

```python
data3=(10)
print(type(data3))
#output:<class'int'>
#why
#because (10) is not a tuple. It is simply an integer 10 enclosed in parenthese.

#python treats it as:
#data3 = 10
```

```python
#how to create a single-element tuple?
#you must use a comma:
data4=(10,)
print(type(data4))
#remember this rule:
#code type
#data = (10) int
#data = (10,) tuple
#data = 10, tuple
#data = (10,20) tuple
```

```python
# tuple indexing
# tuples are ordered and immutable.
data=(10,20,30,40,50)
print(data[2]) # Indexing is used to access inindividual elements of a tuple and Tuple and tuple indexing starts from 0.
print(data[-1]) #Negative indexing starts from -1from the right side.
```

```python
#A Tuple allows duplicate elements.
data=(10,20,30,40,50,30) #Duplicate Elements
print(data)
```

```python
#tuple is immutable
#a tuple is immutable, which means its element cannot be changed, updated, or deleted after creation.
data=(10,20,30,40,50)
data[0]=88 #tries to change the first element from 10 to 88. therefore, python raises a typeerror.

#typeerror: 'tuple' object does not support item assignment
```

```python
#tuple packing:
studentdata="Aman","MCA AI", 78.65 #tuple packing: tuple packing means storing multiple values into a single tuple without explicity using parethensis.
print(type(studentdata))

#tuple unpacking
name, course, avgMarks=studentdata #tuple unpacking means extracting the element of a tuple into separate variables.
print(name)
#Note: The number of variables should normally match the number of tuple elements.
```

```python
record="Amit","MCA",23,True #tuple packing
print(record[0])
print(record[1])
print(record)

name,course,age,att=record
#name,course,age,att="Amit","MCA",23,True  tuple unpacking
print(name)

#unpacking using *
# * in tuple unpacking is called extended unpacking. It allows one variable to collect multiple remaining values.
n,*ab=record
#note: The * operator collets the remaining elements into a list.
print(n)
print(ab)
```

```python
record="Amit","MCA",23,True #tuple packing
name,course,age,att=record #ValueError: too many values to unpack (expacted 3)
#Note: The number of variables should normally match the number of tuple elements.
print(record)
```

```python
record="Amit","MCA",23,True #tuple packing
name,*ab,att=record
print(name)
print(ab)
print(att)
```

```python
#Len() Returns the total number of elements
data=(10,20,30,40,50,30)
print(len(data))

#count() counts how many times a value occurs
print(data.count(30))

print(sum(data)) #Returns the sum of numeric elements
print(max(data)) #returns the largest element
print(min(data)) #returns the smallest element
print(data.index(10)) #returns the index of the first occurrence
```

```python
x,y=10,20
print(x+y)
print(x-y)
print(x*y)
print(x/y)

#floor division
print(7//2) # 3  Floor division returns the floor value (greatest int less than equal to the result)
print(-7//2) # -4 
print(2**3) #8 Exponentiation ** is used to calculate the power of number.

#Important Note : Floor division does not simply remove the decimalpart; it moves the result toward negative infinity.
```

```python
#Compund Assignment Operators: Compound assignment operators are used to perform an operation and assign the result back to the same variable.

x,y=40,50
x+=y # x=x+y >>>   x=40+50    >>>  x=90
print(x)

x-=y # x=x-y >>>  x=90-50    >>> x=40
print(x)

x*=y # x=x*y >>>  x=40*50   >>>  x=2000
print(x)

x/=y # x=x/y >>>  x=2000/50 >>> x=40.0
print(x)
```
