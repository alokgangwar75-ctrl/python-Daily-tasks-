```python
data=[10,34,-61,68,-2,263,-5]
posCount=0
negCount=0
i=0
#print(len(data))
while i<len(data):
    if(data[i]>0):
        posCount+=1
    else:
        negCount+=1
    i+=1
print(posCount)
print(negCount)
```
