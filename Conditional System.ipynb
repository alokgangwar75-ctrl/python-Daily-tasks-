```python
# Conditional Statements.

# User Input and Type castting:
age=int(input("Enter your age:"))
print(type(age))
```

```python
# Simple if:
age=int(input("Enter your age:"))
if age>=18:
    print("Valid for Vote")
```

```python
# If else:
age=int(input("Enter your age:"))
if age>=18:
    print("Valid for vote")
else:
    print("Not Valid for vote")
```

```python
# Even and Odd
num=int(input("Enter a number to check even or odd"))
if num%2==0:
    print("Even")
else:
    print("Odd")
```

```python
# Positive
num=int(input("Enter number to check + or -"))
if num>0:
    print("Number is + :")
else:
    print("Number is - :")
```

```python
# If elif else (elif)
num=int(input("Enter number to check + or -"))
if num==0:
    print("Number is Zero")
elif num>0:
    print("Number is + :")
else:
    print("Numbedr is - :")
```

```python
# Grading system
marks=int(input("Enter your marks:"))
if marks>=90:
    print("Grade A+")
elif marks>=80:
    print("Grade A")
elif marks>=70:
    print("Grade B")
elif marks>=60:
    print("Grade C")
else:
    print("Fail")
```
