```python
employees = {
    "Rahul": {"salary": 60000, "department": "IT"}, 
    "Amit": {"salary": 40000, "department": "HR"}, 
    "Priya": {"salary": 70000, "department": "IT"}, 
    "Neha": {"salary": 45000, "department": "Finance"}
}
for name, data in employees.items():
    salary=data["salary"]
    department=data["department"]
    if(salary>=50000):
        if(department=="IT"):
            print(name," Senior IT Employee")
        elif(department=="HR"):
            print(name," Senior HR Employee")

    else:
        print(name," Senior Employee")
else:
    print(name," Junior Employee")
```

```python
data = {
    "a":10,
    "b":15,
    "c":25,
    "d":30,
    "e":40
}
for key,value in data.items():
    if (value % 2 == 0):
        if (value > 20):
            print(key,"large Even")
        else:
            print(key,"Even")
    else:
        print(key,"odd")
```

```python
students = {
        "Rahul": {"attendance": 85, "marks": 70},
        "Amit": {"attendance": 60, "marks": 80}, 
        "Priya": {"attendance": 90, "marks": 35}, 
        "Neha": {"attendance": 78, "marks": 55}
}
for name, data in students.items():
        attendance = data["attendance"]
        marks = data["marks"]
        if(attendance >= 75):
            if(marks>=40):
                print(name,"Eligible")
            else:
                print(name,"Faild in exam")
        else:
            print(name,"Not Eligible")
```

```python
account = {
    "Rahul":25000,
    "Amit":5000,
    "Priya":0,
    "Neha":-2000
}
for name, balance in account.items():
    if(balance < 0):
        print(name,"overdraft")
    elif(balance == 0):
        print(name,"No balance")
    elif(balance >= 10000):
        print(name,"Premium Account")
    else:
        print(name,"Regular Account")
```

```python
employes = {
    "Rahul": {"salary": 85000, "experience": 7},
    "Amit": {"salary": 35000, "experience": 2},
    "Priya": {"salary": 65000, "experience": 5},
    "Neha": {"salary": 45000, "experience": 3},
    "Rohit": {"salary": 90000, "experience": 8}
}
for name, data in employes.items():
    salary = data["salary"]
    experience = data["experience"]

    if (salary >= 50000):
        if(experience >= 5):
            print(name,"senior")
        else:
            print(name,"mid level")
    else:
        if(experience >= 3):
            print(name,"experience junior")
        else:
            print(name,"fresher")
```

```python
students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}

for name, data in students.items():
    marks = data["marks"]
    attendance = data["attendance"]

    if (attendance >= 75):
        if (marks >= 80):
            print(name, "Distinction")
        elif (marks >= 60):
            print(name, "First Division")
        elif (marks >= 40):
            print(name, "Pass")
        else:
            print(name, "Not Eligible")
    else:
        print(name, "Fail")
```
