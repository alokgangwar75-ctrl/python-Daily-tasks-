```python
data=dict()
print(type(data))
print(data)
```

```python
#Values can be duplicate
data={"name1": "Mohit","age": 23,"att": True,"name2": "Mohit"}
print(data)
```

```python
#Keys can't be duplicate
data={"name": "Mohit","age": 23,"att": True,"name": "Amit"}
print(data)
```

```python
#dict are Mutable
data={"name": "Mohit","age": 23,"att": True}
print(data)

data["name"]="Priya"
print(data)

print(data["name"]) #using Key>>
#print(data["college"]) #using key>>

print(data.get("name"))
print(data.get("college"))
```

```python
#len()
data={"name": "Mohit","age": 23,"att": True}
print(len(data))

#values()
print(data.values())

#items()
print(data.items())
```

```python
#pop()
data={"name": "Mohit","age": 23,"att": True}
data.pop("age")
print(data)
```

```python
#pop()
data={"name": "Mohit","age": 23,"att": True}
data.pop("age")
```

```python
#popitem()
data={"name": "Mohit","age": 23,"att": True}
data.popitem()
```

```python
#clear()
data={"name": "Mohit","age": 23,"att": True}
data.clear()
print(data)
```

```python
#copy()
data={"name": "Mohit","age": 23,"att": True}
newdata=data.copy()
#print(newdata)

data["name"]="Rohit"
print(newdata)
print(data)
```

```python
data={"name": "Mohit","age": 23,"att": True}
print("name" in data)
print("Mohit" in data.values())
```

```python
data={"name": "Mohit","age": 23,"att": True}
print("name" not in data)
print("Mohit" not in data.values())
```
