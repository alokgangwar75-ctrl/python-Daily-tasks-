```python
# Ticket ki original price
ticketPrice = 500

# Check kar rahe hain ki seat available hai ya nahi
seatAvailable = True

# Agar seat available hai tabhi ticket booking proceed hogi
if seatAvailable == True:

    # User se age input le rahe hain
    age = int(input("Enter Your Age: "))

    # 5 saal se chhote bachchon ke liye ticket free hai
    if age < 5:
        print("Ticket is Free")
        print("Final Price: 0")

    # 5 se 12 saal tak ke liye 50% discount
    elif age <= 12:
        print("50% Discount on Ticket")

        # Original price ka 50% discount calculate kar rahe hain
        discount = ticketPrice * 50 / 100

        # Discount ko ticket price se minus kar rahe hain
        ticketPrice -= discount

        # Discount ke baad final price print kar rahe hain
        print("Discount Amount:", discount)
        print("Final Price:", ticketPrice)

    # 60 saal se zyada age ke liye 30% discount
    elif age > 60:
        print("30% Discount on Ticket")

        # Original price ka 30% discount calculate kar rahe hain
        discount = ticketPrice * 30 / 100

        # Discount ko ticket price se minus kar rahe hain
        ticketPrice -= discount

        # Discount ke baad final price print kar rahe hain
        print("Discount Amount:", discount)
        print("Final Price:", ticketPrice)

    # Baaki sabhi ages ke liye full ticket price
    else:
        print("Full Amount")
        print("Final Price:", ticketPrice)

# Agar seat available nahi hai
else:
    print("House Full")
```
