```python
# Relational / comparision operators in Python. Relational operaters are used to compare two values.
# The result of every relational operater is always a Boolean value: True or False.
x=10
y=20

print(x==y) # Equal to == operator checks wheather two values are equal.
print(x<y) # Less Than < operator checks wheather the left-hand value is less than the right-hand value.
print(x>y) # Greater Than > operators checks wheather the left-hand value is greate than the right-hand value.
print(x>=y) # Greater Than or equal to >= operators checks two conditions: 
                                                            # left value is greater than the right value.
                                                            # or both values are eqaul.
print(x<=y) #less than or equalto <= operator checks two conditions;
                                                            # left value is less than the right value.
                                                            # or both values are equal.
print(x!=y) # Not equal to != operator checks wheather two values are different.
```

```python
# Logical and oprerator in python:
# The and operator is used to combine two or more conditions. And = BOTH conditions must be true.
x=10
y=20
print(x<y and y>x)
print(x<y and x==y)
print(x>y and x<y)
print(x>y and x==y)
```

```python
# Membership Operators : in and not in 
# Membership operators are used to check wheather a particular value exits inside a sequence or collection, such as a list, tuple, string, set, etc.

data=[10,20,30,66,88]
print(22 in data) # checks wheather a value exits, it returns either True or False.
print(20 in data)
```

```python
data=[10,20,30,66,88]
print(22 not in data) #checks wheather a value does no exit , It returns either True or False.
print(20 not in data)
```

```python
# == vs is with List
x=[10,20] #Python creates a list object in memory and x refers to that object.
y=[10,20] #Python creates another seprate list object. Although the values are the same, the two lists are different object.
          #Becouse list belongs to Mutable Type.
print(x==y) # Output: True  why? The == opeator compares the values/ content of two objects.
print(x is y) # Output: False Why? The is operater checks objects identity.
x.append(33) #The list referred to by x is modified. Only the x list changes because x and y refer to different list objects.
print(x==y) # Output: False because: x - [10,20,33] y - [10,20] Their contents are now different.
print(id(x))
print(id(y))
```

```python
x=10 # x refers to the int object 10.
y=10 # Python can make y refer to the same int object 10.
print(id(x))
print(id(y))
print(x==y) # Therefore: id(x) == id(y) is typically: True
print(x is y) #And: x is y returns: True because x and y refer to the same object.

# Note : For int and some other immutable objects, Python may reuse objects, but you should generally used == to compare values
# And use is mainly when checking object identity. 
```

```python
x=[10,20] # Python creates a list object in memory.
y=x #Python does not create another list. Instead, y refers to the same list object.
#so, both x and y point to the same object.
print(x is y) #True Why? Because x and y refer to the same list object.
# is operator  checks whether two variable refer to the same  object in memory.
print(id(x))
print(id(y))
print(x==y) # == operator checks value/contemt equality.
```
